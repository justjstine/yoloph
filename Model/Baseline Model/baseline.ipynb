## Create data.yaml

```python
data_yaml = """
path: /kaggle/input/datasets/kurtjustine/pothole-dataset/Combined Dataset

train: train/images
val: val/images

names:
  0: pothole
"""

with open("/kaggle/working/data.yaml", "w") as f:
    f.write(data_yaml)

print("data.yaml created!")
```

## Train Baseline Model

```python
import os

for root, dirs, files in os.walk("/kaggle/input"):
    print(root)
```

```python
from ultralytics import YOLO

model = YOLO("yolov8s-seg.pt")

results = model.train(
    data="/kaggle/working/data.yaml",
    epochs=100
)
```

## Validate the Best Model

```python
best_model = YOLO("/kaggle/working/runs/segment/train/weights/best.pt")

metrics = best_model.val()

print(metrics)
```
